```python
# Convert only the problematic column to numeric
df['Speed Limit[km/h]'] = pd.to_numeric(df['Speed Limit[km/h]'], errors='coerce')

# Now fill missing values
df['Speed Limit[km/h]'].fillna(df['Speed Limit[km/h]'].median(), inplace=True)

# This column is already numeric, just fill
df['Speed Limit with Direction[km/h]'].fillna(
    df['Speed Limit with Direction[km/h]'].median(), inplace=True
)
```

```python
print(df['Speed Limit[km/h]'].dtype)
print(df['Speed Limit with Direction[km/h]'].isnull().sum())
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
target = 'Energy_Consumption'

threshold = 0.7 * len(df)

high_missing_cols = df.isnull().sum()[
    (df.isnull().sum() > threshold) & (df.columns != target)
].index

df.drop(columns=high_missing_cols, inplace=True)

print("Dropped Columns:", list(high_missing_cols))
print("Target still present:", target in df.columns)
```

```python
# Step 2: Interpolate medium-missing time-series features
medium_missing_cols = [
    'Energy_Consumption',
    'Absolute Load[%]',
    'MAF[g/sec]'
]

for col in medium_missing_cols:
    if col in df.columns:
        df[col] = df[col].interpolate(method='linear')
```

```python
# Step 3: Handle low-missing values
df['Gradient'] = df['Gradient'].interpolate(method='linear')
df['Engine RPM[RPM]'] = df['Engine RPM[RPM]'].interpolate(method='linear')
df['Vehicle Speed[km/h]'] = df['Vehicle Speed[km/h]'].interpolate(method='linear')

# Categorical → fill with mode
df['Class of Speed Limit'].fillna(df['Class of Speed Limit'].mode()[0], inplace=True)
```

```python
# Fill any remaining numeric NaNs with median
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())

print("Remaining Missing Values:\n", df.isnull().sum().sum())
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# If Timestamp exists, use it
if 'Timestamp(ms)' in df.columns:
    df['Timestamp'] = pd.to_datetime(df['Timestamp(ms)'], unit='ms')
    df['Hour'] = df['Timestamp'].dt.hour
    df['Day'] = df['Timestamp'].dt.day
```

```python
df['Speed_x_RPM'] = df['Vehicle Speed[km/h]'] * df['Engine RPM[RPM]']
df['Elevation_Change'] = df['Elevation Smoothed[m]'].diff().fillna(0)
```

ENCODE

```python
le = LabelEncoder()

if 'Class of Speed Limit' in df.columns:
    df['Class of Speed Limit'] = le.fit_transform(df['Class of Speed Limit'])

if 'Match Type' in df.columns:
    df['Match Type'] = le.fit_transform(df['Match Type'])
```

```python
#drop leaking colmns
drop_cols = [
    'VehId', 'Trip',
    'Matchted Latitude[deg]', 'Matched Longitude[deg]',
    'Timestamp(ms)', 'Timestamp'
]
df = df.drop(columns=drop_cols, errors='ignore')

# Remove invalid/zero/negative energy values
df = df[df['Energy_Consumption'] > 0]

# Clip extreme outliers (1% to 99%)
q_low = df['Energy_Consumption'].quantile(0.01)
q_high = df['Energy_Consumption'].quantile(0.99)
df['Energy_Consumption'] = df['Energy_Consumption'].clip(q_low, q_high)

# Log transform target to stabilize tiny values
df['Energy_Consumption'] = np.log1p(df['Energy_Consumption'])


X = df.drop(columns=['Energy_Consumption'])
y = df['Energy_Consumption']
```

```python
print('Energy_Consumption' in df.columns)
print(df.columns.tolist())
```

```python
target = 'Energy_Consumption'

X = df.drop(columns=[target])
y = df[target]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)
```

```python
# model = RandomForestRegressor(n_estimators=100, random_state=42)
model = RandomForestRegressor(
    n_estimators=300,
    max_depth=15,
    random_state=42,
    n_jobs=-1
)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("MAE:", mean_absolute_error(y_test, y_pred))
print("R2 Score:", r2_score(y_test, y_pred))
```
