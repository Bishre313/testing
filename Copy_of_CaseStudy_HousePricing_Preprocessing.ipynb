#Preprocessing


#Step1:missing values - Filling and removing

```python
cols = ['Sale Price','No of Bathrooms','Flat Area (in Sqft)','Lot Area (in Sqft)','Area of the House from Basement (in Sqft)']

for col in cols:
    house_pricing[col] = house_pricing[col].fillna(house_pricing[col].median())

house_pricing
```

```python
house_pricing.dropna(subset=['Zipcode','Latitude','Longitude','Living Area after Renovation (in Sqft)'],inplace=True)
# Here removing specific null value row from corresponding column
```

```python
house_pricing.isna().sum()
```

#Step2: Outlier Detection - IQR

```python
num_cols = house_pricing.select_dtypes(include=['int64','float64']).columns
```

```python
for col in num_cols:
    Q1 = house_pricing[col].quantile(0.25)
    Q3 = house_pricing[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = ((house_pricing[col] < lower) | (house_pricing[col] > upper)).sum()

    if outlier_count > 0:
        print(col, "=", outlier_count, "outliers")
        print(f"  Lower Bound: {lower}")
        print(f"  Upper Bound: {upper}")

        # clip function to set the value in range between lower and upper bound
        house_pricing[col] = house_pricing[col].clip(lower=lower, upper=upper)
```

```python
for col in num_cols:
    Q1 = house_pricing[col].quantile(0.25)
    Q3 = house_pricing[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((house_pricing[col] < lower) | (house_pricing[col] > upper)).sum()
    if count > 0:
        print(col, count)
```

### Step3: Encoding & Scailing

```python
# Encode categorical columns
categorical_cols = house_pricing.select_dtypes(include='object').columns
house_pricing = pd.get_dummies(house_pricing, columns=categorical_cols, drop_first=True)
```

```python
house_pricing.info()
```

```python
# Scaling numeric columns
from sklearn.preprocessing import StandardScaler
sscaler = StandardScaler()
house_pricing[num_cols] = sscaler.fit_transform(house_pricing[num_cols])
```
